```python
for i in range(5): # 5 not included
    print(i)
```

```python
for i in range (1,10,3):
    print(i) # range(start, end,step_size)
```

```python
#string iterables
st1 = "Rohit"
for i in st1:
    print(i)
```

```python
#while loop
## continues until the condition is false

count = 0 
while count<5:
    print(count)
    count += 1
```

```python
count = 0
while count %2 == 0:
    print(count)
    count += 1
```

```python
## break and continue
for i in range(10):
    
    if i == 5:
        continue
    print(i)
```

```python
for i in range(10):
    
    if i == 5:
        break
    print(i)
```

```python
# pass
# pass statement is a null operation, it does nothing
for i in range(10):
    if i == 3:
        print("the number is:", i)
        pass
    print(i)
```

```python
inp = int(input("Enter the term: "))
sum = 0 
i = 1
while i <= inp:
    sum = sum + i
    i = i + 1

print(f"The sum upto {inp}th term is: {sum}")


```

```python
# Calculate sum of N natural numbers using while and for loop

num = int(input("Enter upto which term you want the sum: "))
sum = 0
for i in range (1, num+1):
    sum += i
print (f"Sum of natural numbers upto {num}th terms is: {sum}")
```
