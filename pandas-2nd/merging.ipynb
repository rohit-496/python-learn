```python
import pandas as pd
```

```python
#merging is used when we have to join using a common column
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries,on="employee_id")
```

```python
pd.merge(employees,salaries,on="employee_id",how="inner")
#inner means in employees id in both data frames which are common merge those only
#keep common ids in both frames(intersection)
```

```python
pd.merge(employees,salaries,on="employee_id",how="outer")
# outer means merge all but if data not found then show NaN
#keep all ids full outer join(union)
```

```python
pd.merge(employees,salaries,on="employee_id",how="left")
#it merges the dataframe wrt to employees(left) dataframe
```

```python
pd.merge(employees,salaries,on="employee_id",how="right")
# it merges the dataframe wrt to right dataframe(salaries)
```

```python
#concatenation (attach the whole dataframe)
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],axis =1)
```
