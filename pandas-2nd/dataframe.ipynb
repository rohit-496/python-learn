```python
import pandas as pd
```

We can create dataframe using lists as well as dictionaries

```python
data_list = {"Name":["Rohit","Sittal","Rajiv","Sandesh","Anjil"],
"Age":[18,20,22,24,23],
"City":["Butwal","Jungle","Home","College","Jorpati"],
"Salary":[60000,70000,80000,32000,25000]}
```

```python
data = pd.DataFrame(data_list)
data
```

```python
data_lyst = [["rohit","sittal"],
             [18,20],
            ["Butwal","Jungle"],
            [60000,70000]]
contents = ["Name","Age","City","Salary"]
pd.DataFrame(data_lyst,contents)
```
